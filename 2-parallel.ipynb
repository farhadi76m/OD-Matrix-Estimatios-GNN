```python
import xml.etree.ElementTree as ET
import numpy as np
import os

TAZ_FILE    = "grids.taz.xml"
abs_taz    = os.path.abspath(TAZ_FILE)

tree  = ET.parse(abs_taz)
zones = [t.attrib["id"] for t in tree.getroot().findall("taz")]
n     = len(zones)
print(f"  zones: {n}  first 3: {zones[:3]}")

def load_od_from_xml(xml_file="final_od.xml", zones=None):
    """
    Load OD matrix from SUMO od.xml / final_od.xml file
    
    Parameters:
        xml_file: path to the OD XML file
        zones: list of zone ids in correct order (if None, will extract from file)
    
    Returns:
        od_matrix (numpy array), zones (list)
    """
    tree = ET.parse(xml_file)
    root = tree.getroot()
    
    # Extract all tazRelations
    relations = []
    for interval in root.findall("interval"):
        for rel in interval.findall("tazRelation"):
            from_zone = rel.get("from")
            to_zone = rel.get("to")
            count = int(rel.get("count", 0))
            relations.append((from_zone, to_zone, count))
    
    # Get all unique zones
    all_zones = set()
    for f, t, _ in relations:
        all_zones.add(f)
        all_zones.add(t)
    
    zone_list = sorted(all_zones) if zones is None else zones
    n = len(zone_list)
    zone_index = {zone: idx for idx, zone in enumerate(zone_list)}
    
    # Create matrix
    od_matrix = np.zeros((n, n), dtype=int)
    
    for from_zone, to_zone, count in relations:
        if from_zone in zone_index and to_zone in zone_index:
            i = zone_index[from_zone]
            j = zone_index[to_zone]
            od_matrix[i, j] = count
    
    print(f"Loaded OD matrix from {xml_file}")
    print(f"Shape: {od_matrix.shape} | Total trips: {od_matrix.sum()}")
    
    return od_matrix, zone_list


# ========================= USAGE =========================

# Load with existing zones (recommended)
od_matrix, loaded_zones = load_od_from_xml("final_od.xml", zones=zones)

print("Matrix shape:", od_matrix.shape)
print("Sample (top-left 5x5):")
print(od_matrix)

# Or load without passing zones (auto-detects)
# od_matrix, loaded_zones = load_od_from_xml("final_od.xml")
```
